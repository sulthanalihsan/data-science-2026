# **3.5 Implementasi dengan Scikit-Learn**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer

# Memuat dataset contoh (misal: breast cancer) agar X dan y terdefinisi
data = load_breast_cancer()
X, y = data.data, data.target

# Pisahkan data latih dan uji, jaga proporsi kelas (stratify)
X_tr, X_te, y_tr, y_te = train_test_split(
X, y, test_size=0.2, stratify=y, random_state=42)

rf = RandomForestClassifier(
n_estimators=300,
max_features="sqrt",
class_weight="balanced", # menangani data tak seimbang
random_state=42)

rf.fit(X_tr, y_tr)

# Probabilitas kelas positif, bukan sekadar label
proba = rf.predict_proba(X_te)[:, 1]
print("Model training completed successfully.")
print(f"Sample probabilities: {proba[:5]}")
```

# **4.6 Implementasi: SMOTE dan Evaluasi**

```python
from imblearn.over_sampling import SMOTE
from sklearn.metrics import classification_report, roc_auc_score
# SMOTE hanya pada DATA LATIH
sm = SMOTE(random_state=42)
X_res, y_res = sm.fit_resample(X_tr, y_tr)
rf.fit(X_res, y_res)
proba = rf.predict_proba(X_te)[:, 1]
pred = (proba >= 0.5).astype(int)
# Evaluasi pada data uji asli
print(classification_report(y_te, pred))
print("ROC-AUC:", roc_auc_score(y_te, proba))
```
